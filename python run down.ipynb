```python
class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return self.price * self.quantity

    def __str__(self):
        return f"Product: {self.name} | Price: {self.price} | Quantity: {self.quantity} | Total Value: {self.total_value()}"


class PerishableProduct(Product):

    def __init__(self, name, price, quantity, expiration_days):
        super().__init__(name, price, quantity)
        # Unique extra attribute for this subclass
        self.expiration_days = expiration_days


class Inventory:

    def __init__(self):
        self._products = []

    def add_product(self, product):  # Fixed parameter name to lowercase 'product'
        """Adds a Product to the inventory."""
        self._products.append(product)

    def total_inventory_value(self):
        """Calculates total inventory value by summing the value of each product."""
        return sum(product.total_value() for product in self._products)


# Instantiate inventory
inventory = Inventory()

# Mix of Product and PerishableProduct instances
inventory.add_product(Product("Phone", 120000.0, 2))
inventory.add_product(PerishableProduct("Milk", 3500, 10, 7))
inventory.add_product(Product("Mouse", 2500.0, 5))
inventory.add_product(PerishableProduct("Bread", 1000, 15, 3))

# Total inventory value using the single abstract method call
grand_total = inventory.total_inventory_value()

print(f"Grand Total Inventory Value: {grand_total:,} FCFA")
```

simulating a few sales, and (6) prints a final formatted report summarizing total value, lowest-stock item, and highestvalue item. There is no single correct structure — the goal is combining the pieces cleanly, not matching a template.

```python
import csv

# Base Class and Inheritance
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = float(price)        # Convert to float
        self.quantity = int(quantity)    # Convert to int

    def total_value(self):
        return self.price * self.quantity

    def record_sale(self, amount):
        if amount <= self.quantity:
            self.quantity -= amount
            print(f"Sold {amount} units of {self.name}.")
        else:
            print(f"Not enough stock to sell {amount} units of {self.name}.")

    def __str__(self):
        return f"{self.name} - Price: {self.price:,.2f} FCFA | Stock: {self.quantity}"

class PerishableProduct(Product):
    """Specialized product type with an expiry date feature."""
    def __init__(self, name, price, quantity, expiry_date):
        super().__init__(name, price, quantity)
        self.expiry_date = expiry_date

    def __str__(self):
        return super().__str__() + f", Expires: {self.expiry_date}"


def load_inventory(filename):
    inventory = []
    try:
        with open(filename, mode='r') as file:
            reader = csv.DictReader(file)
            for row in reader:
                if row.get('expiry_date'):
                    p = PerishableProduct(row['name'], row['price'], row['quantity'], row['expiry_date'])
                else:
                    p = Product(row['name'], row['price'], row['quantity'])
                inventory.append(p)
        print(f"Successfully loaded inventory from '{filename}'.")
    except FileNotFoundError:
        print(f"Warning: '{filename}' not found. Initializing with default stock.")
        # Default fallback data if file is missing
        inventory = [
            PerishableProduct("Milk", 1500, 20, "2026-08-25"),
            Product("Rice (5kg)", 6500, 15),
            Product("Cooking Oil", 4500, 4),
            PerishableProduct("Bread", 500, 5, "2026-08-20")
        ]
    return inventory

def save_inventory(filename, inventory):
    with open(filename, mode='w', newline='') as file:
        fieldnames = ['name', 'price', 'quantity', 'expiry_date']
        writer = csv.DictWriter(file, fieldnames=fieldnames)
        writer.writeheader()
        for p in inventory:
            writer.writerow({
                'name': p.name,
                'price': p.price,
                'quantity': p.quantity,
                'expiry_date': getattr(p, 'expiry_date', '')
            })
    print(f"Updated inventory saved to '{filename}'.\n")


# Reporting with *args and **kwargs
def print_header(title, *notes, **metadata):
    print("=" * 45)
    print(f"  {title.upper()}")
    print("=" * 45)
    for note in notes:
        print(f"Note: {note}")
    for key, value in metadata.items():
        print(f"{key.replace('_', ' ').title()}: {value}")
    print("-" * 45)


# Main Execution
def main():
    filename = "inventory.csv"
    
    # Load starting inventory
    inventory = load_inventory(filename)

    # Simulate some sales
    print("\n--- Simulating Daily Sales ---")
    inventory[0].record_sale(3)
    inventory[2].record_sale(2)
    print("------------------------------\n")

    # Using filter(), map(), and lambda
    reorder_threshold = 5
    low_stock = list(filter(lambda p: p.quantity < reorder_threshold, inventory))

    # Calculate total value of each product
    product_values = list(map(lambda p: p.total_value(), inventory))
    grand_total_value = sum(product_values)

    # Key metrics calculation
    lowest_stock_item = min(inventory, key=lambda p: p.quantity)
    highest_value_item = max(inventory, key=lambda p: p.total_value())

    # Save updated stock back to CSV
    save_inventory(filename, inventory)

    # Formatted Output Report
    print_header(
        "Nkwen Traders Stock Report", 
        "Automated end-of-day summary"
    )

    print("CURRENT INVENTORY:")
    for item in inventory:
        print(f"  * {item}")

    print("\nREORDER ALERTS (< 5 units):")
    if low_stock:
        for item in low_stock:
            print(f"  ! {item.name} (Only {item.quantity} left)")
    else:
        print("  All stock levels healthy.")

    print("\nMETRICS SUMMARY:")
    print(f"  * Grand Total Stock Value : {grand_total_value:,.2f} FCFA")
    print(f"  * Lowest Stock Item       : {lowest_stock_item.name} ({lowest_stock_item.quantity} units)")
    print(f"  * Highest Value Item      : {highest_value_item.name} ({highest_value_item.total_value():,.2f} FCFA)")
    print("=" * 45)

if __name__ == "__main__":
    main()
```
