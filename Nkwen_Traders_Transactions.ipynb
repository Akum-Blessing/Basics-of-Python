```python
def total_revenue(transactions):
    """Add up quantity * price for every transaction in the list."""
    total = 0
    for t in transactions:
        total = total + (t["quantity"] * t["price"])
    return total


# Test it right away, on its own
print("Total weekly revenue:", total_revenue(sales_data), "FCFA")
```

```python
def best_selling_product(transactions):
    """Return the (product, total_quantity) with the highest quantity sold."""

    # Step A: add up quantity sold per product
    totals = {}
    for t in transactions:
        product = t["product"]
        if product in totals:
            totals[product] = totals[product] + t["quantity"]
        else:
            totals[product] = t["quantity"]


    # Step B: walk through the keys, and look up each value manually
    best_product = None
    best_quantity = 0
    for product in totals:
        quantity = totals[product]
        if quantity > best_quantity:
            best_product = product
            best_quantity = quantity

    return best_product, best_quantity

data=(best_selling_product(sales_data))
print(data)
```

```python
def revenue_by_day(transactions):
    """Return a dictionary of {day: total_revenue_for_that_day}."""
    totals = {}
    for t in transactions:
        day = t["day"]
        revenue = t["quantity"] * t["price"]
        if day in totals:
            totals[day] = totals[day] + revenue
        else:
            totals[day] = revenue
    return totals


# Test it on its own
daily = revenue_by_day(sales_data)
for day, revenue in daily.items():
    print(f"{day} - {revenue} FCFA")
    
```

```python
def sort_days_by_revenue(daily_totals):
    """Return a list of (day, revenue) pairs, sorted highest revenue first."""
    
    # Build our own copy, one entry at a time, instead of using dict()
    remaining = {}
    for day in daily_totals:
        revenue = daily_totals[day]
        remaining[day] = revenue
    return remaining

    result = []
    
    while len(remaining) > 0:
        best_day = None
        best_revenue = 0
        for day in remaining:
            revenue = remaining[day]
            if revenue > best_revenue:
                best_day = day
                best_revenue = revenue
        
        result.append((best_day, best_revenue))
        del remaining[best_day]
    
    return result

print(sort_days_by_revenue(revenue_by_day(sales_data)))
```

```python
def print_report(transactions):
    """Calculate everything and print a readable weekly report."""
    total = total_revenue(transactions)
    product, qty = best_selling_product(transactions)
    daily = revenue_by_day(transactions)
    ranked_days = sort_days_by_revenue(daily)

    print("=== Traders Weekly Sales Report ===")
    print(f"Total Revenue: {total:,} FCFA")
    print(f"Best-Selling Product: {product} ({qty} units)")
    print()
    print(f"-----Day by day transactions----- ")
    for day, revenue in ranked_days.items():
        print(f"  {day}: {revenue:,} FCFA")
    print(f"Days Ranked by Revenue (best to worst) ---- {sort_days_by_revenue(revenue_by_day(sales_data))}:")
```

```python
def main():
    print_report(sales_data)


if __name__ == "__main__":
    main()
```
