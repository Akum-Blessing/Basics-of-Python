1. Read the CSV file you just created back into Python as a list of dictionaries (without pandas).
2. Loop through the re-loaded data and print a formatted inventory report from it, the same way you did with the original in-memory list — confirm the output matches.

```python
import csv

products = [
    {"name": "Rice", "price": 25000, "quantity": 1},
    {"name": "Onions", "price": 3500, "quantity": 12},
    {"name": "Salt", "price": 4000, "quantity": 11},
    {"name": "Bread", "price": 1000, "quantity": 13},
    {"name": "Beans", "price": 8000, "quantity": 6},
]

filename = "inventory.csv"
fieldnames = ["name", "price", "quantity"]

# Writing to the CSV file
with open(filename, "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(products)

print("Data successfully written to inventory.csv!\n")

# Reading back from the CSV file
with open("inventory.csv", "r") as file:
    reader = csv.DictReader(file)
    reloaded_products = []

    for row in reader:
        # Cast numerical strings back to numbers
        product = {
            "name": row["name"],
            "price": float(row["price"]),
            "quantity": int(row["quantity"]),
        }

        # Single quotes inside f-string double quotes
        print(f"Name: {product['name']}")
        print(f"Price: {product['price']} XAF")
        print(f"Quantity: {product['quantity']}")
        print("_" * 50)

        reloaded_products.append(product)
```

1. Write a function (or standalone block, if you haven't covered functions yet) that attempts to open and read a CSV filename.
2. Manually corrupt one row in your CSV (e.g., delete a value or insert a non-numeric price), then update your Step 6 reader to catch and report that row's error — using ValueError or a manual field check — without crashing or skipping the rest of the file
3. Wrap the file-reading logic in try/except/finally so that:
o if the file exists, it processes and reports the data,
o if the file does not exist, it prints a clear error message instead of crashing,
o either way, a finally message prints confirming the attempt is complete.
4. Test this twice: once with the real filename (should succeed), and once with a deliberately misspelled filename (should fail gracefully). Confirm both runs behave as expected and that you can see the difference in your printed output.


```python
import csv

products = [
    {"name": "Rice", "price": 25000, "quantity": 1},
    {"name": "Onions", "price": 3500, "quantity": 12},
    {"name": "Salt", "price": 4000, "quantity": 11},
    {"name": "Bread", "price": 1000, "quantity": 13},
    {"name": "Beans", "price": 8000, "quantity": 6},
]

filename = "inventory.csv"
fieldnames = ["name", "price", "quantity"]

# Writing to the CSV file
with open(filename, "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(products)

print("Data successfully written to inventory.csv!\n")

# Reading back from the CSV file
with open("inventory.csv", "r") as file:
    reader = csv.DictReader(file)
    reloaded_products = []

    for row in reader:
        # Cast numerical strings back to numbers
        product = {
            "name": row["name"],
            "price": float(row["price"]),
            "quantity": int(row["quantity"]),
        }

        # Single quotes inside f-string double quotes
        print(f"Name: {product['name']}")
        print(f"Price: {product['price']} XAF")
        print(f"Quantity: {product['quantity']}")
        print("_" * 50)

        reloaded_products.append(product)


def process_inventory_file(filename: str) -> list[dict]:
    reloaded_products = []
    print(f"Attempting to process '{filename}' ---")

    try:
        with open(filename, "r") as file:
            reader = csv.DictReader(file)

            for line_num, row in enumerate(reader, start=2):  # start=2 accounts for header row
                try:
                    # Validate missing fields manually
                    if not row["name"] or not row["price"] or not row["quantity"]:
                        raise ValueError("Missing essential field values.")

                    # Attempt type conversion (catches non-numeric prices/quantities)
                    product = {
                        "name": row["name"],
                        "price": float(row["price"]),
                        "quantity": int(row["quantity"]),
                    }

                    reloaded_products.append(product)
                    print(
                        f"Successfully loaded: {product['name']} "
                        f"({product['quantity']} @ {product['price']} XAF)"
                    )

                except ValueError as row_err:
                    print(
                        f"Row {line_num} Error: Corrupted data in row {dict(row)}. "
                        f"Reason: {row_err}. Skipping row."
                    )

    except FileNotFoundError:
        print(f"File Error: The file '{filename}' was not found.")

    finally:
        print(f"Completed file processing attempt for '{filename}' ---\n")

    return reloaded_products
    
 # SETUP CORRUPTED FILE FOR TESTING
# Writing a test CSV containing 1 corrupted price ('invalid_price') and 1 missing value
corrupted_data = [
    {"name": "Rice", "price": 25000, "quantity": 1},
    {"name": "Onions", "price": "invalid_price", "quantity": 12},  # Non-numeric price
    {"name": "Salt", "price": 4000, "quantity": 11},
    {"name": "", "price": 1000, "quantity": 13},  # Missing name
    {"name": "Beans", "price": 8000, "quantity": 6},
]

with open("inventory_corrupted.csv", "w") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "price", "quantity"])
    writer.writeheader()
    writer.writerows(corrupted_data)

# Handles missing/bad rows gracefully, loads valid rows, prints finally statement
valid_items = process_inventory_file("inventory_corrupted.csv")
print(f"Total valid products loaded: {len(valid_items)}\n")

# Catches FileNotFoundError gracefully and prints finally statement
invalid_items = process_inventory_file("non_existent_inventory.csv")
```

The outer except block caught FileNotFoundError, which previously crashed the program with FileNotFoundError: No such file or directory: 'wrong_filename.csv'. The inner except block caught ValueError, which previously stopped execution on bad rows with messages like ValueError: could not convert string to float: 'invalid_price'.

Without wrapping the missing file read in a try/except block, Python would have raised an unhandled FileNotFoundError as soon as it tried to execute open("wrong_filename.csv"). This unhandled exception would immediately terminate the entire script, preventing any code after that line from running.
